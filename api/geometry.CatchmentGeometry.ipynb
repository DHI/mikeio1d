# geometry.CatchmentGeometry

``` python
geometry.CatchmentGeometry(self, points)
```

A utility class for working with catchment geometries.

## Parameters

| Name | Type | Description | Default |
|-----|---------------|----------------------------------------------|-------|
| points | list\[tuple\[float, float\]\] | List of points (x, y) defining the catchment boundary. The first and last points should be the same. | *required* |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [from_res1d_catchment](#mikeio1d.geometry.CatchmentGeometry.from_res1d_catchment) | Create a CatchmentGeometry from an IRes1DCatchment object. |
| [to_shapely](#mikeio1d.geometry.CatchmentGeometry.to_shapely) | Convert to a shapely geometry. |

### from_res1d_catchment

``` python
geometry.CatchmentGeometry.from_res1d_catchment(res1d_catchment)
```
