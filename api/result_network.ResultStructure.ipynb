# result_network.ResultStructure

``` python
result_network.ResultStructure(self, structure_id, reach, data_items, res1d)
```

Class for wrapping a list of ResultData structure data items belonging
to the same structure.

## Parameters

| Name | Type | Description | Default |
|---------|----------------|----------------------------------------|--------|
| structure_id | str | Structure ID. | *required* |
| reach | IRes1DReach | Reach where the structure belongs to. | *required* |
| data_items | list of IDataItem objects. | A list of MIKE 1D IDataItem objects corresponding to a given structure. | *required* |
| res1d | Res1D | Res1D object the reach belongs to. | *required* |

## Attributes

| Name | Type | Description |
|----------------|--------|------------------------------------------------|
| data_items_dict | dict | A dictionary from quantity id to a data item. |
| chainage | float | Chainage where the structure is located on the reach. |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [add_query](#mikeio1d.result_network.ResultStructure.add_query) | Add a query to ResultNetwork.queries list. |
| [get_m1d_dataset](#mikeio1d.result_network.ResultStructure.get_m1d_dataset) | Get IRes1DDataSet object associated with ResultStructure. |
| [get_query](#mikeio1d.result_network.ResultStructure.get_query) | Get a QueryDataStructure for given data item. |
| [read](#mikeio1d.result_network.ResultStructure.read) | Read the time series data for all quantities at this location into a DataFrame. |
| [to_dataframe](#mikeio1d.result_network.ResultStructure.to_dataframe) | Read the time series data for all quantities at this location into a DataFrame. |

### add_query

``` python
result_network.ResultStructure.add_query(data_item)
```

Add a query to ResultNetwork.queries list.

### get_m1d_dataset

``` python
result_network.ResultStructure.get_m1d_dataset(m1d_dataitem=None)
```

Get IRes1DDataSet object associated with ResultStructure.

This is the reach IRes1DDataSet object because ResultStructure objects
do not have a corresonding IRes1DDataSet object.

#### Parameters

| Name         | Type      | Description                  | Default |
|--------------|-----------|------------------------------|---------|
| m1d_dataitem | IDataItem | Ignored for ResultStructure. | `None`  |

#### Returns

| Name | Type          | Description                                           |
|--------|---------------|-------------------------------------------------|
|      | IRes1DDataSet | IRes1DDataSet object associated with ResultStructure. |

### get_query

``` python
result_network.ResultStructure.get_query(data_item)
```

Get a QueryDataStructure for given data item.

#### Returns

| Name | Type               | Description                                |
|------|--------------------|--------------------------------------------|
|      | QueryDataStructure | Query for the data item on this structure. |

### read

``` python
result_network.ResultStructure.read(column_mode=None)
```

Read the time series data for all quantities at this location into a
DataFrame.

#### Parameters

| Name | Type | Description | Default |
|-----|-----|-----------------------------------------------------------|----|
| column_mode | str(optional) | Specifies the type of column index of returned DataFrame. ‘all’ - column MultiIndex with levels matching TimeSeriesId objects. ‘compact’ - same as ‘all’, but removes levels with default values. ‘timeseries’ - column index of TimeSeriesId objects | `None` |

#### Returns

| Name | Type         | Description                                           |
|---------|--------------|--------------------------------------------------|
|      | pd.DataFrame | Time series data for all quantities at this location. |

### to_dataframe

``` python
result_network.ResultStructure.to_dataframe(column_mode=None)
```
