# result_network.ResultStructure

``` python
result_network.ResultStructure(self, structure_id, reach, data_items, res1d)
```

Class for wrapping a list of ResultData structure data items belonging
to the same structure.

## Parameters

| Name | Type | Description | Default |
|---------|----------------|----------------------------------------|--------|
| structure_id | str | Structure ID. | *required* |
| reach | IRes1DReach | Reach where the structure belongs to. | *required* |
| data_items | list of IDataItem objects. | A list of MIKE 1D IDataItem objects corresponding to a given structure. | *required* |
| res1d | Res1D | Res1D object the reach belongs to. | *required* |

## Attributes

| Name | Type | Description |
|----------------|--------|------------------------------------------------|
| data_items_dict | dict | A dictionary from quantity id to a data item. |
| chainage | float | Chainage where the structure is located on the reach. |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [add](#mikeio1d.result_network.ResultStructure.add) | Add quantities at this location to ResultNetwork.queue for reading later. |
| [add_query](#mikeio1d.result_network.ResultStructure.add_query) | Add a query to ResultNetwork.queries list. |
| [get_m1d_dataset](#mikeio1d.result_network.ResultStructure.get_m1d_dataset) | Get IRes1DDataSet object associated with ResultStructure. |
| [get_query](#mikeio1d.result_network.ResultStructure.get_query) | Get a QueryDataStructure for given data item. |
| [plot](#mikeio1d.result_network.ResultStructure.plot) | Plot the time series data for quantities at this location. |
| [read](#mikeio1d.result_network.ResultStructure.read) | Read the time series data for quantities at this location into a DataFrame. |
| [to_csv](#mikeio1d.result_network.ResultStructure.to_csv) | Extract time series data for quantities at this location into a csv file. |
| [to_dataframe](#mikeio1d.result_network.ResultStructure.to_dataframe) | Read the time series data for quantities at this location into a DataFrame. |
| [to_dfs0](#mikeio1d.result_network.ResultStructure.to_dfs0) | Extract time series data for quantities at this location into a dfs0 file. |
| [to_txt](#mikeio1d.result_network.ResultStructure.to_txt) | Extract time series data for quantities at this location into a txt file. |

### add

``` python
result_network.ResultStructure.add(quantities=None)
```

### read

``` python
result_network.ResultStructure.read(column_mode=None, *, quantities=None)
```

Read the time series data for quantities at this location into a
DataFrame.

#### Parameters

| Name | Type | Description | Default |
|----|--------|---------------------------------------------------------|----|
| column_mode | str(optional) | Specifies the type of column index of returned DataFrame. ‘all’ - column MultiIndex with levels matching TimeSeriesId objects. ‘compact’ - same as ‘all’, but removes levels with default values. ‘timeseries’ - column index of TimeSeriesId objects | `None` |
| quantities | str \| list\[str\] \| None | Quantity id or list of quantity ids to read. If None, all quantities are read. | `None` |

#### Returns

| Name | Type         | Description                                       |
|------|--------------|---------------------------------------------------|
|      | pd.DataFrame | Time series data with one column per time series. |

#### Raises

| Name | Type       | Description                                      |
|------|------------|--------------------------------------------------|
|      | ValueError | If a quantity is not available at this location. |

#### Returns

| Name | Type         | Description                                           |
|---------|--------------|--------------------------------------------------|
|      | pd.DataFrame | Time series data for all quantities at this location. |

### to_csv

``` python
result_network.ResultStructure.to_csv(
    file_path,
    time_step_skipping_number=1,
    *,
    quantities=None,
)
```
