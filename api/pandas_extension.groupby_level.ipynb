# pandas_extension.groupby_level

``` python
pandas_extension.groupby_level(df, level_name)
```

Group DataFrame for aggregations along the specific level name.

## Parameters

| Name | Type | Description | Default |
|-----------|------------|---------------------------------------|-----------|
| df | pd.DataFrame | DataFrame with hierarchical column index. | *required* |
| level_name | str | Name of level in the hierarchical column index. | *required* |

## Returns

| Name | Type         | Description        |
|------|--------------|--------------------|
| df   | pd.DataFrame | Grouped DataFrame. |

## Examples

``` python
>>> df.m1d.groupby_level("duplicate").first()
Out[1]: The DataFrame with the first value for each duplicate.
```
