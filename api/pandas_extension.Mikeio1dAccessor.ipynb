# pandas_extension.Mikeio1dAccessor

``` python
pandas_extension.Mikeio1dAccessor(self, pandas_obj)
```

Uses Pandas Extension API to register a custom accessor for DataFrames.

More information can be found here:

https://pandas.pydata.org/docs/development/extending.html#registering-custom-accessors

The accessor provides convenience methods for working with DataFrames
with a MultiIndex. The intent is as a facade for various helper
functions that live elsewhere in the package.

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [agg_chainage](#mikeio1d.pandas_extension.Mikeio1dAccessor.agg_chainage) | Wrap ResultReaches.agg_chainage. |
| [compact](#mikeio1d.pandas_extension.Mikeio1dAccessor.compact) | Wrap compact_dataframe. |
| [groupby](#mikeio1d.pandas_extension.Mikeio1dAccessor.groupby) | Wrap pd.DataFrame.groupby. The groupby is performed on the columns of the DataFrame, which are in the form of a MultiIndex. |
| [groupby_chainage](#mikeio1d.pandas_extension.Mikeio1dAccessor.groupby_chainage) | Wrap pd.DataFrame.groupby. The groupby is performed on the columns of the DataFrame, which are in the form of a MultiIndex. |
| [groupby_level](#mikeio1d.pandas_extension.Mikeio1dAccessor.groupby_level) | Wrap groupby_level. The groupby is performed on the columns of the DataFrame, which are in the form of a MultiIndex. |
| [query](#mikeio1d.pandas_extension.Mikeio1dAccessor.query) | Wrap pd.DataFrame.query. The query is performed on the columns of the DataFrame, which are in the form of a MultiIndex. |

### agg_chainage

``` python
pandas_extension.Mikeio1dAccessor.agg_chainage(agg=None)
```
