# pandas_extension.groupby_chainage

``` python
pandas_extension.groupby_chainage(df, **kwargs)
```

Group results for aggregation along the chainage axis.

## Parameters

| Name | Type | Description | Default |
|------|---------|--------------------------------------------------|--------|
| df | pd.DataFrame | DataFrame with results. Must have hierarchical column index (e.g. column_mode = ‘all’). | *required* |

## Returns

| Name | Type | Description |
|---------|------------------|----------------------------------------------|
| groupby | TransposedGroupBy | GroupBy object, which can be used for aggregation. |
