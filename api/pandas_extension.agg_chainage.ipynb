# pandas_extension.agg_chainage

``` python
pandas_extension.agg_chainage(df, agg=None, gb_kwargs=None, **kwargs)
```

Aggregate results along the chainage axis.

## Parameters

| Name | Type | Description | Default |
|-----|----------------|--------------------------------------------|-------|
| df | pd.DataFrame | DataFrame with results. Must have hierarchical column index (e.g. column_mode = ‘all’). | *required* |
| agg | (function, str, list or dict) | Aggregation function(s) to apply. Same as pandas.DataFrame.agg. | `None` |

## Returns

| Name | Type         | Description                        |
|------|--------------|------------------------------------|
| df   | pd.DataFrame | DataFrame with aggregated results. |
