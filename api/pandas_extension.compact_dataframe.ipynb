# pandas_extension.compact_dataframe

``` python
pandas_extension.compact_dataframe(df)
```

Convert a DataFrame with a hierarchical column index to a compact
DataFrame.

A compact DataFrame removes levels where every value matches the
TimeSeriesId default value.

## Parameters

| Name | Type         | Description                               | Default    |
|------|--------------|-------------------------------------------|------------|
| df   | pd.DataFrame | DataFrame with hierarchical column index. | *required* |

## Returns

| Name | Type         | Description        |
|------|--------------|--------------------|
| df   | pd.DataFrame | Compact DataFrame. |
