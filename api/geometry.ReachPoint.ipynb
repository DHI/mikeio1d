# geometry.ReachPoint

``` python
geometry.ReachPoint(self, point_type, chainage, x, y, z)
```

A utility class for working with points along a reach.

## Parameters

| Name | Type | Description | Default |
|---------|------------|------------------------------------------|---------|
| point_type | ReachPointType | Either DIGIPOINT or GRIDPOINT | *required* |
| chainage | float | Chainage along the reach. Need not match geometric distance. | *required* |
| x | float | X coordinate | *required* |
| y | float | Y coordinate | *required* |
| z | float | Z coordinate | *required* |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [from_digipoint](#mikeio1d.geometry.ReachPoint.from_digipoint) | Create a ReachPoint from an IRes1DDigiPoint object. |
| [from_gridpoint](#mikeio1d.geometry.ReachPoint.from_gridpoint) | Create a ReachPoint from an IRes1DGridPoint object. |
| [is_digipoint](#mikeio1d.geometry.ReachPoint.is_digipoint) | Check if the point is a digipoint. |
| [is_gridpoint](#mikeio1d.geometry.ReachPoint.is_gridpoint) | Check if the point is a gridpoint. |
| [to_shapely](#mikeio1d.geometry.ReachPoint.to_shapely) | Convert to a Shapely Point. |

### from_digipoint

``` python
geometry.ReachPoint.from_digipoint(res1d_digipoint)
```
