# result_network.ResultGlobalData

``` python
result_network.ResultGlobalData(self, data_item, global_datas, res1d)
```

Class for wrapping ResultData global data items.

By itself it is also a dict, which contains mapping between data item
quantity ID and IDataItem object.

## Parameters

| Name | Type | Description | Default |
|-----------|-------------------|----------------------------------|---------|
| data_item | IDataItem | MIKE 1D IDataItem object for a global data item. | *required* |
| global_datas | ResultGlobalDatas object. | A wrapper object for all global data items. | *required* |
| res1d | Res1D | Res1D object the global data belongs to. | *required* |

## Attributes

| Name | Description |
|------------------------------------|------------------------------------|
| [derived_quantities](#mikeio1d.result_network.ResultGlobalData.derived_quantities) | A list of available derived quantities. |
| [group](#mikeio1d.result_network.ResultGlobalData.group) | The TimeSeriesIdGroup assosciated with this location. |
| [quantities](#mikeio1d.result_network.ResultGlobalData.quantities) | A list of available quantities. |
| [res1d](#mikeio1d.result_network.ResultGlobalData.res1d) | The Res1D instance that this location belongs to. |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [add](#mikeio1d.result_network.ResultGlobalData.add) | Add quantities at this location to ResultNetwork.queue for reading later. |
| [add_query](#mikeio1d.result_network.ResultGlobalData.add_query) | Add a query to ResultNetwork.queries list. |
| [get_m1d_dataset](#mikeio1d.result_network.ResultGlobalData.get_m1d_dataset) | Get IRes1DDataSet object associated with ResultGlobalData. |
| [get_query](#mikeio1d.result_network.ResultGlobalData.get_query) | Get a QueryDataGlobal for given data item. |
| [plot](#mikeio1d.result_network.ResultGlobalData.plot) | Plot the time series data for quantities at this location. |
| [read](#mikeio1d.result_network.ResultGlobalData.read) | Read the time series data for quantities at this location into a DataFrame. |
| [to_csv](#mikeio1d.result_network.ResultGlobalData.to_csv) | Extract time series data for quantities at this location into a csv file. |
| [to_dataframe](#mikeio1d.result_network.ResultGlobalData.to_dataframe) | Read the time series data for quantities at this location into a DataFrame. |
| [to_dfs0](#mikeio1d.result_network.ResultGlobalData.to_dfs0) | Extract time series data for quantities at this location into a dfs0 file. |
| [to_txt](#mikeio1d.result_network.ResultGlobalData.to_txt) | Extract time series data for quantities at this location into a txt file. |

### add

``` python
result_network.ResultGlobalData.add(quantities=None)
```

### read

``` python
result_network.ResultGlobalData.read(column_mode=None, *, quantities=None)
```

Read the time series data for quantities at this location into a
DataFrame.

#### Parameters

| Name | Type | Description | Default |
|----|--------|---------------------------------------------------------|----|
| column_mode | str(optional) | Specifies the type of column index of returned DataFrame. ‘all’ - column MultiIndex with levels matching TimeSeriesId objects. ‘compact’ - same as ‘all’, but removes levels with default values. ‘timeseries’ - column index of TimeSeriesId objects | `None` |
| quantities | str \| list\[str\] \| None | Quantity id or list of quantity ids to read. If None, all quantities are read. | `None` |

#### Returns

| Name | Type         | Description                                       |
|------|--------------|---------------------------------------------------|
|      | pd.DataFrame | Time series data with one column per time series. |

#### Raises

| Name | Type       | Description                                      |
|------|------------|--------------------------------------------------|
|      | ValueError | If a quantity is not available at this location. |

#### Returns

| Name | Type         | Description                                           |
|---------|--------------|--------------------------------------------------|
|      | pd.DataFrame | Time series data for all quantities at this location. |

### to_csv

``` python
result_network.ResultGlobalData.to_csv(
    file_path,
    time_step_skipping_number=1,
    *,
    quantities=None,
)
```
