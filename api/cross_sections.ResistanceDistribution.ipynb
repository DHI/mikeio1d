# cross_sections.ResistanceDistribution

``` python
cross_sections.ResistanceDistribution()
```

The type of resistance distribution used in the cross section.

This is a simple wrapper of DHI.Mike1D.Generic.ResistanceDistribution.

## Attributes

| Name | Description |
|------------------------------------|------------------------------------|
| [CONSTANT](#mikeio1d.cross_sections.ResistanceDistribution.CONSTANT) | Resistance value is constant for all water depths. Processed resistance values are not used. |
| [DISTRIBUTED](#mikeio1d.cross_sections.ResistanceDistribution.DISTRIBUTED) | Fully user-distributed along cross section profile. Resistance values are part of processed data. |
| [EXPONENT_VARYING](#mikeio1d.cross_sections.ResistanceDistribution.EXPONENT_VARYING) | Resistance is depth dependent with an exponential variation between top and bottom. |
| [UNIFORM](#mikeio1d.cross_sections.ResistanceDistribution.UNIFORM) | Uniform within the regular channel and vegetation parts, resp. (ie, 2 values). Resistance values are |
| [ZONES](#mikeio1d.cross_sections.ResistanceDistribution.ZONES) | High/low flow zones along the cross section profile, defined by low flow markers (4, LeftLowFlowBank |
