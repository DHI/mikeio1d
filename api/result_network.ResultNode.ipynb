# result_network.ResultNode


### read

``` python
result_network.ResultNode.read(column_mode=None, *, quantities=None)
```

Read the time series data for quantities at this location into a
DataFrame.

#### Parameters

| Name | Type | Description | Default |
|----|--------|---------------------------------------------------------|----|
| column_mode | str(optional) | Specifies the type of column index of returned DataFrame. ‘all’ - column MultiIndex with levels matching TimeSeriesId objects. ‘compact’ - same as ‘all’, but removes levels with default values. ‘timeseries’ - column index of TimeSeriesId objects | `None` |
| quantities | str \| list\[str\] \| None | Quantity id or list of quantity ids to read. If None, all quantities are read. | `None` |

#### Returns

| Name | Type         | Description                                       |
|------|--------------|---------------------------------------------------|
|      | pd.DataFrame | Time series data with one column per time series. |

#### Raises

| Name | Type       | Description                                      |
|------|------------|--------------------------------------------------|
|      | ValueError | If a quantity is not available at this location. |

#### Returns

| Name | Type         | Description                                           |
|---------|--------------|--------------------------------------------------|
|      | pd.DataFrame | Time series data for all quantities at this location. |

### to_csv

``` python
result_network.ResultNode.to_csv(
    file_path,
    time_step_skipping_number=1,
    *,
    quantities=None,
)
```
