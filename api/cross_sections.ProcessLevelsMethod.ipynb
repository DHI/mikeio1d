# cross_sections.ProcessLevelsMethod

``` python
cross_sections.ProcessLevelsMethod()
```

How to generate processing levels.

This is a simple wrapper of DHI.Mike1D.Generic.ProcessingOption

## Attributes

| Name | Description |
|------------------------------------|------------------------------------|
| [AUTOMATIC_LEVELS](#mikeio1d.cross_sections.ProcessLevelsMethod.AUTOMATIC_LEVELS) | Use the built-in algorithm to find appropriate processed levels. |
| [EQUIDISTANT_LEVELS](#mikeio1d.cross_sections.ProcessLevelsMethod.EQUIDISTANT_LEVELS) | Equidistant processed levels between a given min/max pair. |
| [USER_DEFINED_LEVELS](#mikeio1d.cross_sections.ProcessLevelsMethod.USER_DEFINED_LEVELS) | Processed levels are user-defined. |
