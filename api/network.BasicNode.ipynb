# network.BasicNode

``` python
network.BasicNode(self, id, data)
```

Concrete :class:`NetworkNode` for programmatic network construction.

## Parameters

| Name | Type | Description | Default |
|--------|------------|-------------------------------------------|-----------|
| id | str | Unique node identifier. | *required* |
| data | pd.DataFrame | Time-indexed DataFrame with one column per quantity. | *required* |

## Examples

``` python
>>> import pandas as pd
>>> time = pd.date_range("2020", periods=3, freq="h")
>>> node = BasicNode("junction_1", pd.DataFrame({"WaterLevel": [1.0, 1.1, 1.2]}, index=time))
```
