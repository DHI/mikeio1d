# cross_sections.Marker

``` python
cross_sections.Marker()
```

The standard markers used in cross sections.

## Notes

User markers are integers greater than 7. They are specified as an
integer, not an enum.

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [from_pretty](#mikeio1d.cross_sections.Marker.from_pretty) | Parse a string created from Marker.pretty to the corresponding marker value. |
| [is_default_marker](#mikeio1d.cross_sections.Marker.is_default_marker) | Check if the int/Marker is a default marker (e.g. left levee bank). |
| [is_user_marker](#mikeio1d.cross_sections.Marker.is_user_marker) | Check if the int/Marker is a user marker (i.e. value \> 7). |
| [list_from_string](#mikeio1d.cross_sections.Marker.list_from_string) | Convert a string of comma-separated markers to a list of integers. |
| [pretty](#mikeio1d.cross_sections.Marker.pretty) | Get a pretty string presentation of the marker. |

### from_pretty

``` python
cross_sections.Marker.from_pretty(marker)
```

Parse a string created from Marker.pretty to the corresponding marker
value.

### is_default_marker

``` python
cross_sections.Marker.is_default_marker(marker)
```

Check if the int/Marker is a default marker (e.g. left levee bank).

### is_user_marker

``` python
cross_sections.Marker.is_user_marker(marker)
```

Check if the int/Marker is a user marker (i.e. value \> 7).

### list_from_string

``` python
cross_sections.Marker.list_from_string(s)
```

Convert a string of comma-separated markers to a list of integers.

### pretty

``` python
cross_sections.Marker.pretty(marker)
```

Get a pretty string presentation of the marker.
