# cross_sections.ResistanceType

``` python
cross_sections.ResistanceType()
```

The type of resistance used in the cross section.

This is a simple wrapper of DHI.Mike1D.Generic.ResistanceFormulation

## Attributes

| Name | Description |
|------------------------------------|------------------------------------|
| [CHEZY](#mikeio1d.cross_sections.ResistanceType.CHEZY) | Chezy. |
| [COLEBROOK_WHITE](#mikeio1d.cross_sections.ResistanceType.COLEBROOK_WHITE) | Colebrook White. Only used with circular CrossSections |
| [DARCY_WEISBACH](#mikeio1d.cross_sections.ResistanceType.DARCY_WEISBACH) | Darcy-Weisbach resistance. |
| [HAZEN_WILLIAMS](#mikeio1d.cross_sections.ResistanceType.HAZEN_WILLIAMS) | Hazen Williams. |
| [MANNINGS_M](#mikeio1d.cross_sections.ResistanceType.MANNINGS_M) | Manning’s M. |
| [MANNINGS_M_RELATIVE](#mikeio1d.cross_sections.ResistanceType.MANNINGS_M_RELATIVE) | Manning’s M, relative, meaning that raw/processed resistance numbers are relative to the local manning |
| [MANNINGS_N](#mikeio1d.cross_sections.ResistanceType.MANNINGS_N) | Manning’s n. |
| [RELATIVE](#mikeio1d.cross_sections.ResistanceType.RELATIVE) | Relative factor. The factor is relative to the globally provided resistance formulation and number. |
